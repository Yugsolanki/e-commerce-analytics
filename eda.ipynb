```python
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine('postgresql+psycopg2://admin:admin@localhost:5432/ecommerce')
```

```python
customer_df = pd.read_csv('./data/archive/olist_customers_dataset.csv')
geolocation_df = pd.read_csv('./data/archive/olist_geolocation_dataset.csv')
order_items_df = pd.read_csv('./data/archive/olist_order_items_dataset.csv')
order_payments_df = pd.read_csv('./data/archive/olist_order_payments_dataset.csv')
order_reviews_df = pd.read_csv('./data/archive/olist_order_reviews_dataset.csv')
orders_df = pd.read_csv('./data/archive/olist_orders_dataset.csv')
products_df = pd.read_csv('./data/archive/olist_products_dataset.csv')
sellers_df = pd.read_csv('./data/archive/olist_sellers_dataset.csv')
product_category_translation_df = pd.read_csv('./data/archive/product_category_name_translation.csv')
```

```python
date_columns = [
    'shipping_limit_date'
]

order_items_df[date_columns] = order_items_df[date_columns].apply(
    pd.to_datetime,
    errors='coerce'
)

order_items_df.info()
```

```python
order_payments_df.info()
print("=================================================================")
order_payments_df.dropna()
```

```python
order_reviews_df.info()
print("=================================================================")
order_reviews_df.dropna()
```

```python
date_columns = [
    'review_creation_date',
    'review_answer_timestamp'
]

order_reviews_df[date_columns] = order_reviews_df[date_columns].apply(
    pd.to_datetime,
    errors='coerce'
)

order_reviews_df.info()
```

```python
orders_df.info()
print("=================================================================")
orders_df.dropna()
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

orders_df[date_columns] = orders_df[date_columns].apply(
    pd.to_datetime,
    errors='coerce'
)

orders_df.info()
```

```python
products_df.info()
print("=================================================================")
products_df.dropna()
```

```python
product_category_translation_df.info()
print("=================================================================")
product_category_translation_df.dropna()
```

```python
with open('./data/import_csv_postgres/create_tables.sql') as file:
    sql_script = file.read()
    
with engine.begin() as connection:
    connection.execute(text(sql_script))
```

```python
customer_df.to_sql(
    'raw_customers',
    engine,
    if_exists='append',
    index=False
)

geolocation_df.to_sql(
    'raw_geolocation',
    engine,
    if_exists='append',
    index=False
)

orders_df.to_sql(
    'raw_orders',
    engine,
    if_exists='append',
    index=False
)

order_items_df.to_sql(
    'raw_order_items',
    engine,
    if_exists='append',
    index=False
)

order_payments_df.to_sql(
    'raw_order_payments',
    engine,
    if_exists='append',
    index=False
)

order_reviews_df.to_sql(
    'raw_order_reviews',
    engine,
    if_exists='append',
    index=False
)

products_df.to_sql(
    'raw_products',
    engine,
    if_exists='append',
    index=False
)

sellers_df.to_sql(
    'raw_sellers',
    engine,
    if_exists='append',
    index=False
)

product_category_translation_df.to_sql(
    'raw_product_category_translation',
    engine,
    if_exists='append',
    index=False
)
```
